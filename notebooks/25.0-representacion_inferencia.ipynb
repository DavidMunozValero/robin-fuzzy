# Resultados de una inferencia en WIFE modelos TSK

##**MODELO**

R0::: IF ((P is cheap) | (P is medium)) THEN 25

R1::: IF ((O is Renfe) | (O is Iryo)) & (P is cheap) THEN 35

R2::: IF (O is Oigo) & ((C is 0) | (C is 1)) THEN 40


## Resultados con traza

Cuando el parámetro *trace* de *model.loop_tsk_inference* se pone a *True* se genera como salida un diccionario similar al siguiente.

```python
{ # Representa la traza de una inferencia del modelo de arriba ante la entrada ['Renfe', 15, 1.0]
    'variables_order': ['O', 'P', 'C'], # Orden de las variables en la entrada
    'input': ['Renfe', 15, 1.0], # Valor de la entrada
    'result': 42.5, # Valor de salida final, en este caso, con Modelo TSK Acumulativo
    # Dentro de esta clave se encuentra una lista de diccionarios, cada una corresponde
    # a la traza de una regla
    'trace_of_each_rule': [
            { # Diccionario de la regla R0: IF ((P is cheap) | (P is medium)) THEN 25
                'name': 'R0', # Nombre de la regla
                'antecedent': { # Información de la ejecución en el antecedente
                    'operators': [], # Solo tiene una PDC dentro de otra PDC debido a los paréntesis que engloban todo
                    'operants':
                    [
                        {'operators': ['|'], # PDC compuesta de dos AP: (P is cheap) | (P is medium)
                         'operants':
                            [
                                { # PA: (P is cheap)
                                    'operator': None,
                                    'variable': 'P',
                                    'input_value': 15, # segundo valor dado que P es la segunda
                                    'function': 'trapezoidal', # Nombre de la función utilizada para ejecutar
                                    'setname_and_values': ('cheap', [0, 0, 10, 20]), # Conjunto difuso más valores del mismo
                                    'result': 0.5 # Resultado final: 15 en el trapezoide [0, 0, 10, 20] da 0.5 de pertenencia
                                },
                                {
                                    'operator': None,
                                    'variable': 'P',
                                    'input_value': 15, 'function': 'trapezoidal',
                                    'setname_and_values': ('medium', [10, 20, 40, 60]),
                                    'result': 0.5
                                }
                            ],
                         'result': 1.0 # Resultado final al aplicar los operandos: 0.5 de cheap + 0.5 de medium igual a 1.0
                        }
                    ],
                    'result': 1.0 # Resultado del antecedente completo
                },
                'consequent': {'term': 25.0}, # Consecuente, en TSK solo es el termino de salida
                'result': 25.0
            },
            {
                'name': 'R1',
                'input': ['Renfe', 15, 1.0],
                'antecedent': {
                    'operators': ['&'],
                    'operants':
                     [
                         {
                            'operators': ['|'],
                            'operants':
                             [
                                 {'operator': None, 'variable': 'O', 'input_value': 'Renfe', 'function': 'no_function', 'setname_and_values': 'Renfe', 'result': 1.0},
                                 {'operator': None, 'variable': 'O', 'input_value': 'Renfe', 'function': 'no_function', 'setname_and_values': 'Iryo', 'result': 0.0}
                            ],
                            'result': 1.0
                        },
                        {
                            'operator': None,
                            'variable': 'P',
                            'input_value': 15,
                            'function': 'trapezoidal',
                            'setname_and_values': ('cheap', [0, 0, 10, 20]),
                            'result': 0.5}
                    ],
                    'result': 0.5
                },
                'consequent': {'term': 35.0},
                'result': 17.5
            }
        ]
}
```
